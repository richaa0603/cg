```python
import json
import sys
from pathlib import Path

ROOT = Path('/content/gitlas/ai_ml')
sys.path.append(str(ROOT))

from src.pipeline import run_pipeline
from src.search_engine import search_projects
```

```python
pipeline_result = run_pipeline(
    seed_query='attendance management system',
    requirement_query='attendance management system',
    max_repos=1000
)
pipeline_result['dataset_count'], pipeline_result['index_size']
```

```python
queries = [
    'attendance management system',
    'insurance claims portal',
    'consulting delivery platform',
    'project management tool'
]

all_results = {}
for q in queries:
    all_results[q] = search_projects(q, top_k=10)

print(json.dumps(all_results, indent=2))
```

```python
for q, repos in all_results.items():
    print(f'\nQuery: {q}')
    for i, repo in enumerate(repos, start=1):
      name = repo.get('repository', '')
      score = repo.get('score', 0)
      language = repo.get('language', 'Unknown')
      stars = repo.get('stars', 0)
      print(f'{i:02d}. {name} | score={score} | {language} | stars={stars}')
```
